```python
import numpy as np
import torch
from dpf2.ai.training import load_numpy_dataset, train_torch_model
from dpf2.ai import TorchSurrogateModel
from dpf2.metadata import Metadata
```

```python
x = np.linspace(-1, 1, 100, dtype=np.float32).reshape(-1, 1)
y = 3 * x + 1
loader = load_numpy_dataset(x, y, batch_size=10)
net = torch.nn.Sequential(torch.nn.Linear(1, 1))
meta = Metadata.with_defaults()
train_torch_model(net, loader, epochs=50, lr=0.1, metadata=meta)
torch.jit.script(net).save('linear.pt')
sur = TorchSurrogateModel.load('linear.pt')
sur.predict(x[:5])
```
