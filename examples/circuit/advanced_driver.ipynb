# Advanced Driver
Demonstrates multi-section RLC blocks, transmission-line segments and crowbar stages.

```python
from dpf2.circuit import TransmissionLineSegment, CrowbarStage
from dpf2.rlc_solver import solve_distributed_circuit
segments = [
    TransmissionLineSegment(0,1,length=1.0,L_per_m=1e-6,R_per_m=0.0,C_per_m=0.0),
    TransmissionLineSegment(1,2,length=1.0,L_per_m=0.0,R_per_m=0.0,C_per_m=1e-6),
]
crowbar = CrowbarStage(0,2,resistance=1e-3,trigger_time=1e-6)
res = solve_distributed_circuit(segments,[crowbar],V0=1e3,t_end=2e-6,dt=1e-7)
res.current[:5]
```
