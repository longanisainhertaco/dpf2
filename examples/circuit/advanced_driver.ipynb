# Advanced Circuit Driver
Demonstrates transmission lines, crowbar stages and statistical switch jitter.

```
from dpf2.circuit_solver import run_circuit_simulation
from dpf2.circuit_config import CircuitConfig, SegmentConfig, SwitchConfig, CrowbarStageConfig

cfg = CircuitConfig.with_defaults().model_copy(update={
    'segments': [SegmentConfig(length=1.0, L=1.0, R=0.0, C=0.0, from_node=0, to_node=1)],
    'switches': [SwitchConfig(from_node=0, to_node=1, closed=False, trigger_times=[50.0], r_on=1.0, r_off=1e6)],
    'crowbar_stages': [CrowbarStageConfig(resistance=0.1, trigger=100.0, jitter=5.0)],
    'trigger_jitter_stddev': 5.0,
    'L_ext': 1.0,
    'R_ext': 1.0,
    'C_ext': 1.0,
    'V0': 1.0,
    'switch_delay': 0.0,
})

t, I, V, _, _ = run_circuit_simulation(cfg, t_end=0.2, num_points=200)
I[:5]
```
