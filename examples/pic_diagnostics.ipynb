```python
import numpy as np
from types import SimpleNamespace
from dpf2.simulation.pic_solver import PICSolver
from dpf2.simulation.utils import FieldManager

cfg = SimpleNamespace(grid_shape=(4,4,4), grid_spacing=(1.0,1.0,1.0), max_dt=None,
                      electromag='yee', boundary_conditions={'x':'reflecting','y':'reflecting','z':'reflecting'},
                      dt=0.1, use_warpx=False, unity_params={}, vdf_bins=8, max_vel=1.0,
                      subgrid_resolution=1, amr=False, density_threshold=1.0,
                      enable_quantum=False, enable_radiation=False, enable_mesh_adaptivity=False)
fm = FieldManager(grid_shape=(4,4,4), dx=1.0, dy=1.0, dz=1.0, domain_lo=(0.0,0.0,0.0), boundary_conditions={})
solver = PICSolver(cfg, fm)
solver.collisions = []
solver.add_species('e', -1.0, 1.0, np.zeros((100,3)), np.random.randn(100,3))
```

```python
edges, hist = solver.compute_energy_spectra('e')
xedges, vedges, H = solver.compute_phase_space('e')
print('Energy bins:', edges)
print('Histogram:', hist)
```
