# Neutron spectra comparison

This notebook compares synthetic spectra with simple reference data.

```python
from dpf2.diagnostics.neutron_spectra import synthetic_tof_spectrum, angular_spectrum, anisotropy_metric, DetectorLayout
import numpy as np

energies = [1.0, 2.0]
flux = [1.0, 1.0]
time_bins = [0.0, 1.0]
tof = synthetic_tof_spectrum(energies, flux, 1.0, time_bins)
reference_tof = [1.0]
print("ToF match:", np.allclose(tof, reference_tof))

layout = DetectorLayout(angles=[0.0, 90.0, 180.0], distance_m=1.0)
spectrum = angular_spectrum(layout.angles_deg(), base_yield=1.0, anisotropy=1.0)
reference = [2.0, 1.0, 0.0]
print("Angular match:", np.allclose(spectrum, reference))
print("Anisotropy metric:", anisotropy_metric(spectrum))
```
