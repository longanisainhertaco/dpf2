# Mass/Current Calibration Demo
This notebook demonstrates using `dpf2.uq.calibration.emcee_calibrate_mass_current` to
estimate posterior distributions of mass and current scaling factors
from time-of-flight and current waveform data.

```python
import numpy as np
from dpf2.uq.calibration import emcee_calibrate_mass_current

rng = np.random.default_rng(0)
time = np.linspace(0, 1e-6, 50)
current_sim = 1e5 * np.sin(2 * np.pi * time / time[-1])
tof_sim = np.array([100e-9])

true_mass = 1.2
true_current = 0.8

current_data = true_current * current_sim + rng.normal(0, 1e4, size=current_sim.shape)
tof_data = true_mass * tof_sim + rng.normal(0, 5e-9, size=tof_sim.shape)

samples = emcee_calibrate_mass_current(
    current_sim,
    current_data,
    tof_sim,
    tof_data,
    bounds={"mass_factor": (0.5, 1.5), "current_factor": (0.5, 1.5)},
    n_steps=500,
    seed=123,
)
print({k: (np.mean(v), np.std(v)) for k, v in samples.items()})
```
