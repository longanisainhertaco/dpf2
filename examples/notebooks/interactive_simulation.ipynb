# Interactive Simulation Demo
Run a DPF simulation with live updating current/voltage plots.

```python
%matplotlib notebook
from dpf2.core.config import DPFConfig
from dpf2.core.simulation import DPFSimulation
import matplotlib.pyplot as plt

cfg = DPFConfig.with_defaults()
sim = DPFSimulation(cfg)

times, currents, voltages = [], [], []
fig, ax = plt.subplots()
(line_i,) = ax.plot([], [], label='current')
(line_v,) = ax.plot([], [], label='voltage')
ax.set_xlabel('time [s]')
ax.legend()

def update(step, time):
    times.append(time)
    currents.append(sim.current)
    voltages.append(sim.voltage)
    line_i.set_data(times, currents)
    line_v.set_data(times, voltages)
    ax.relim(); ax.autoscale_view()

sim.run(progress_cb=update)
```
