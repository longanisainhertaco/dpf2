# Anisotropy Analysis
This notebook demonstrates computing beam-target and thermal yield anisotropy using `dpf2.diagnostics`.

```python
from dpf2.diagnostics import IonBeamEDF, yield_components_with_anisotropy, angular_yield_map

class SimpleBeam(IonBeamEDF):
    def energy_distribution(self, angle_deg):
        return [1.0, 2.0], [1.0, 1.0]

beam = SimpleBeam()
angles = [0.0, 90.0]
cross_section = lambda e: e
distance = 1.0
time_bins = [0.0, 1.0]
reactivity = [1e-20, 1e-20]
ion_density = [1e19, 1e19]
dt = 1e-6

result = yield_components_with_anisotropy(
    beam,
    cross_section,
    angles,
    distance,
    time_bins,
    reactivity,
    ion_density,
    dt,
)
result

energy_bins = [0.0, 1.0, 2.0]
angular_map = angular_yield_map(beam, cross_section, angles, energy_bins)
angular_map
```
