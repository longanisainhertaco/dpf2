# Bayesian Calibration
Sample posterior distributions for a simple model.

```python
import numpy as np
from dpf2.uq.calibration import bayesian_calibration

def model(p):
    return np.array([p[0] + p[1]])

data = np.array([1.0])
bounds = {"a": (0.0, 1.0), "b": (0.0, 1.0)}
posterior = bayesian_calibration(model, bounds, data, n_samples=100, seed=0)
posterior
```
