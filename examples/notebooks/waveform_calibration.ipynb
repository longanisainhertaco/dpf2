# Waveform Calibration
Demonstrate calibration of mass and current scaling factors from a noisy current waveform using `emcee_calibrate_waveform`.

```python
import numpy as np
from dpf2.uq.calibration import emcee_calibrate_waveform

rng = np.random.default_rng(0)
time_sim = np.linspace(0, 1e-6, 50)
current_sim = 1e5 * np.sin(2 * np.pi * time_sim / time_sim[-1])

true_mass = 1.1
true_current = 0.9
time_data = time_sim
scaled_time = true_mass * time_sim
current_data = true_current * np.interp(time_data, scaled_time, current_sim, left=0.0, right=0.0)
current_data += rng.normal(0, 1e4, size=current_data.shape)

samples = emcee_calibrate_waveform(
    time_sim,
    current_sim,
    time_data,
    current_data,
    bounds={"mass_factor": (0.5, 1.5), "current_factor": (0.5, 1.5)},
    n_steps=500,
    seed=123,
)
print({k: (float(np.mean(v)), float(np.std(v))) for k, v in samples.items()})
```
