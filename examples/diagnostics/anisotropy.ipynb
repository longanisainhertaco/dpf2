# Anisotropy Diagnostics

This example demonstrates computing directional yields and generating a
time-of-flight vs. I–V spike correlation report.

```python
from dpf2.synthetic_diagnostics import directional_yields, autocorrelated_tof_iv_report
from dpf2.core.bases import CouplingState

history = [CouplingState(current=c, voltage=c) for c in [0.0, 1.0, 5.0, 1.0, 0.0]]
dt = 1e-9
report_path = autocorrelated_tof_iv_report(history, dt, 1.0, energies_mev=[2.45])
directional = directional_yields(100.0, 1e18, 1e20)
directional, report_path
```
