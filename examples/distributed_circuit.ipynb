# Distributed Circuit Example
This notebook compares a lumped RLC circuit with a simple two-segment distributed network.

```python
import sys, runpy
sys.path.append('../src')
runpy.run_path('../tests/numpy_stub.py')
from dpf2.circuit_config import CircuitConfig
from dpf2.circuit_solver import run_circuit_simulation

cfg_lumped = CircuitConfig.with_defaults()
t_l, i_l, v_l, *_ = run_circuit_simulation(cfg_lumped, t_end=5.0)

cfg_dist = cfg_lumped.model_copy(update={
    'segments': [
        {'length': 1.0, 'L': cfg_lumped.L_ext/2, 'R': cfg_lumped.R_ext/2, 'C': cfg_lumped.C_ext/2},
        {'length': 1.0, 'L': cfg_lumped.L_ext/2, 'R': cfg_lumped.R_ext/2, 'C': cfg_lumped.C_ext/2}
    ],
    'switches': [{'closed': True}, {'closed': True}]
})
t_d, i_d, v_d, *_ = run_circuit_simulation(cfg_dist, t_end=5.0)

print('lumped current sample:', i_l[:5])
print('distributed current sample:', i_d[:5])
```
